R syntax drives me up the wall, so here I'm just formatting the Fst tables that were generated in STAMPP:

```python
import pandas as pd
import numpy as np
```

```python
fst = pd.read_csv("../stagdb_paper_fst_matrix_boot.txt", sep = " ")
```

```python
fst.loc[fst["Population2"].str.contains("Mexico, "), "Population2"] = "Mexico1, GULF"
fst.loc[fst["Population2"] == "Mexico", "Population2"] = "Mexico2"
fst.loc[fst["Population1"].str.contains("Mexico, "), "Population1"] = "Mexico1, GULF"
fst.loc[fst["Population1"] == "Mexico", "Population1"] = "Mexico2"
```

```python
fst["Fst"] = fst["Fst"].round(3)
```

```python
fsttable = fst.sort_values("Population1").pivot(index='Population2', 
                                     columns='Population1', values='Fst')
```

```python
fsttable.rename(index={"Mexico1, GULF": 'Mexico, Gulf Side', 
                       "Mexico2": 'Mexico'}, inplace = True)
fsttable.rename(columns={"Mexico1, GULF": 'Mexico, Gulf Side', 
                       "Mexico2": 'Mexico'}, inplace = True)
```

```python
missing = pd.DataFrame(data = [[np.nan] * len(fsttable.columns)], columns = fsttable.columns)
missing.rename(index={0: 'Bahamas'}, inplace = True)
```

```python
fsttable = pd.concat([missing, fsttable])
```

```python
fsttable["USVI"] = np.nan
```

```python
fst["CI"] = "[" + fst["Lower bound CI limit"].round(3).astype(str) + ",\n" + fst["Upper bound CI limit"].round(3).astype(str) + "]"
```

```python
fst.loc[fst["Population2"] == "Mexico1, GULF", "Population2"] = "Mexico, Gulf Side"
fst.loc[fst["Population1"] == "Mexico1, GULF", "Population1"] = "Mexico, Gulf Side"
fst.loc[fst["Population2"] == "Mexico2", "Population2"] = "Mexico"
fst.loc[fst["Population1"] == "Mexico2", "Population1"] = "Mexico"
```

```python
for column in fsttable.columns:
    for index in fsttable.index:
        if column != index:
            try:
                fsttable[column][index] = (fst["CI"][fst["Population2"] == column]
                                           [fst["Population1"] == index].reset_index(drop=True)[0]) 
            except:
                pass
```
