```python
import pandas as pd
import numpy as np
```

```python
%%bash
source ~/.bashrc
conda activate orbicella
bcftools query -l ../filtered_calls/allsamples_lifted_to_jaAcrPala1.3_conffiltered_indvmissfiltered_downsampled_maffiltered_ldfiltered.vcf.gz \
    > ../final_sample_set_for_popgen.txt
```

```python
samples = pd.read_csv("../final_sample_set_for_popgen.txt", header = None, names = ["Affymetrix ID"])
```

```python
meta = pd.read_csv("../curated_metadata.tsv", sep = "\t")
```

```python
samples = samples.merge(meta, how = "left", on = "Affymetrix ID")
```

```python
for region in samples.Region.unique().tolist():
    samples[["Affymetrix ID"]][samples["Region"] == region].to_csv("../" + region.replace(" ", "_") + ".txt", 
                                                                   header = None, index = False)
```
